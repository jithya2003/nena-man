```python
import os
import numpy as np
import pandas as pd
import librosa
from tqdm.auto import tqdm

base = "/content/drive/MyDrive/Behavior Detection/EmoReact_V_1.0/EmoReact_V_1.0"
data_root = os.path.join(base, "Data")

split_df = pd.read_csv(
    os.path.join(base, "emoreact_audio_split.csv")
)

emotion_cols = [
    "curiosity",
    "uncertainty",
    "excitement",
    "happiness",
    "surprise",
    "disgust",
    "fear",
    "frustration"
]

# Merge labels with split information
label_df = df.copy()
label_df["video_id"] = label_df["video_id"].astype(str).str.strip()

meta = split_df.merge(
    label_df[["video_id"] + emotion_cols],
    on="video_id",
    how="inner"
)

output_dir = os.path.join(base, "audio_features")
os.makedirs(output_dir, exist_ok=True)

results = []

for _, row in tqdm(meta.iterrows(), total=len(meta), desc="Extracting audio"):

    video_id = row["video_id"]
    split = row["split"]

    # Find video in Train folder
    video_path = os.path.join(
        data_root,
        "Train",
        video_id + ".mp4"
    )

    if not os.path.exists(video_path):
        print("Missing:", video_id)
        continue

    try:
        audio, sr = librosa.load(
            video_path,
            sr=16000,
            mono=True
        )

        # MFCC
        mfcc = librosa.feature.mfcc(
            y=audio,
            sr=sr,
            n_mfcc=40
        )

        # Log-Mel
        mel = librosa.feature.melspectrogram(
            y=audio,
            sr=sr,
            n_mels=64
        )

        log_mel = librosa.power_to_db(
            mel,
            ref=np.max
        )

        # Temporal statistics
        feature_vector = np.concatenate([
            mfcc.mean(axis=1),
            mfcc.std(axis=1),
            log_mel.mean(axis=1),
            log_mel.std(axis=1)
        ]).astype(np.float32)

        results.append([
            video_id,
            split,
            *feature_vector,
            *[row[c] for c in emotion_cols]
        ])

    except Exception as e:
        print("ERROR:", video_id, e)

feature_cols = (
    [f"mfcc_mean_{i}" for i in range(40)] +
    [f"mfcc_std_{i}" for i in range(40)] +
    [f"mel_mean_{i}" for i in range(64)] +
    [f"mel_std_{i}" for i in range(64)]
)

columns = (
    ["video_id", "split"] +
    feature_cols +
    emotion_cols
)

audio_df = pd.DataFrame(results, columns=columns)

output_file = os.path.join(
    output_dir,
    "emoreact_audio_features.csv"
)

audio_df.to_csv(output_file, index=False)

print("\n===== EXTRACTION COMPLETE =====")
print("Videos extracted:", len(audio_df))
print("Feature columns:", len(feature_cols))
print("Output:", output_file)
print("\nSplit counts:")
print(audio_df["split"].value_counts())
```

```python
# AUDIO MODEL - PREPARE TRAINING DATA

import numpy as np
import pandas as pd

df_audio = pd.read_csv(audio_file)

# Target
y = (df_audio["frustration"] > 0).astype(int).to_numpy()

# Audio feature columns only
exclude_cols = {
    "video_id",
    "split",
    "label",
    "behaviour",
    "state",
    "audio_frustration"
}

feature_cols = [
    c for c in df_audio.columns
    if c not in exclude_cols
    and pd.api.types.is_numeric_dtype(df_audio[c])
    and c not in [
        "curiosity",
        "uncertainty",
        "excitement",
        "happiness",
        "surprise",
        "disgust",
        "fear",
        "frustration",
        "valence"
    ]
]

X = df_audio[feature_cols].to_numpy(dtype=np.float32)

# Clean invalid values
X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Number of features:", len(feature_cols))
print("Classes:", np.bincount(y))
print("NaN:", np.isnan(X).sum())
print("Inf:", np.isinf(X).sum())
```

```python
# AUDIO MODEL - TRAIN / VALIDATION / TEST SPLIT

from sklearn.preprocessing import StandardScaler
import numpy as np

train_mask = df_audio["split"] == "train"
val_mask   = df_audio["split"] == "validation"
test_mask  = df_audio["split"] == "test"

X_train = X[train_mask.to_numpy()]
y_train = y[train_mask.to_numpy()]

X_val = X[val_mask.to_numpy()]
y_val = y[val_mask.to_numpy()]

X_test = X[test_mask.to_numpy()]
y_test = y[test_mask.to_numpy()]

# Fit scaler ONLY on training data
audio_scaler = StandardScaler()

X_train = audio_scaler.fit_transform(X_train)
X_val   = audio_scaler.transform(X_val)
X_test  = audio_scaler.transform(X_test)

print("TRAIN:", X_train.shape, "Classes:", np.bincount(y_train))
print("VAL  :", X_val.shape,   "Classes:", np.bincount(y_val))
print("TEST :", X_test.shape,  "Classes:", np.bincount(y_test))
```

```python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# Convert to tensors
Xtr = torch.tensor(X_train, dtype=torch.float32)
ytr = torch.tensor(y_train, dtype=torch.long)

Xv = torch.tensor(X_val, dtype=torch.float32)
yv = torch.tensor(y_val, dtype=torch.long)

Xt = torch.tensor(X_test, dtype=torch.float32)
yt = torch.tensor(y_test, dtype=torch.long)

train_loader = DataLoader(
    TensorDataset(Xtr, ytr),
    batch_size=32,
    shuffle=True
)

# Class weights: inverse frequency
class_counts = np.bincount(y_train)
class_weights = len(y_train) / (2.0 * class_counts)

weights = torch.tensor(
    class_weights,
    dtype=torch.float32
).to(device)

print("Class weights:", class_weights)


class AudioBehaviourNet(nn.Module):
    def __init__(self, input_size=208):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(input_size, 128),
            nn.ReLU(),
            nn.Dropout(0.30),

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.25),

            nn.Linear(64, 2)
        )

    def forward(self, x):
        return self.network(x)


model = AudioBehaviourNet(input_size=X_train.shape[1]).to(device)

criterion = nn.CrossEntropyLoss(weight=weights)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001,
    weight_decay=1e-4
)

print(model)
```

```python
# IMPROVED AUDIO MODEL SETUP
# - Balanced sampling for Frustrated class
# - Weight decay to reduce overfitting

from torch.utils.data import WeightedRandomSampler

# Class counts
class_counts = torch.bincount(torch.tensor(y_train, dtype=torch.long))
print("Original class counts:", class_counts.tolist())

# Give more sampling probability to minority class
class_weights = 1.0 / class_counts.float()

sample_weights = class_weights[
    torch.tensor(y_train, dtype=torch.long)
]

sampler = WeightedRandomSampler(
    weights=sample_weights,
    num_samples=len(sample_weights),
    replacement=True
)

# Balanced training loader
train_loader = DataLoader(
    TensorDataset(
        torch.tensor(X_train, dtype=torch.float32),
        torch.tensor(y_train, dtype=torch.long)
    ),
    batch_size=32,
    sampler=sampler
)

# Validation loader
val_loader = DataLoader(
    TensorDataset(
        torch.tensor(X_val, dtype=torch.float32),
        torch.tensor(y_val, dtype=torch.long)
    ),
    batch_size=32,
    shuffle=False
)

# Fresh model
model = AudioBehaviourNet(
    input_size=X_train.shape[1]
).to(device)

# No extra class-weighting here because sampler already balances classes
criterion = torch.nn.CrossEntropyLoss()

# Weight decay helps reduce overfitting
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001,
    weight_decay=1e-4
)

print("Improved training setup ready.")
```

```python
import copy
from sklearn.metrics import f1_score

best_val_f1 = 0.0
best_state_improved = None
patience = 5
patience_counter = 0

EPOCHS = 30

for epoch in range(EPOCHS):

    # ===== TRAIN =====
    model.train()
    train_correct = 0
    train_total = 0

    for X_batch, y_batch in train_loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        optimizer.zero_grad()

        output = model(X_batch)
        loss = criterion(output, y_batch)

        loss.backward()
        optimizer.step()

        pred = output.argmax(dim=1)

        train_correct += (pred == y_batch).sum().item()
        train_total += y_batch.size(0)

    train_acc = train_correct / train_total

    # ===== VALIDATION =====
    model.eval()
    val_preds = []
    val_true = []

    with torch.no_grad():
        for X_batch, y_batch in val_loader:
            X_batch = X_batch.to(device)

            output = model(X_batch)
            pred = output.argmax(dim=1)

            val_preds.extend(pred.cpu().numpy())
            val_true.extend(y_batch.numpy())

    val_acc = (np.array(val_preds) == np.array(val_true)).mean()

    val_f1 = f1_score(
        val_true,
        val_preds,
        average="macro",
        zero_division=0
    )

    print(
        f"Epoch {epoch+1:02d}/{EPOCHS} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val Macro-F1: {val_f1:.4f}"
    )

    # ===== EARLY STOPPING =====
    if val_f1 > best_val_f1:
        best_val_f1 = val_f1
        best_state_improved = copy.deepcopy(model.state_dict())
        patience_counter = 0
    else:
        patience_counter += 1

    if patience_counter >= patience:
        print("\nEarly stopping triggered.")
        break

print("\nBest Improved Validation Macro-F1:", round(best_val_f1, 4))
```

```python
import librosa
import numpy as np
import pandas as pd

audio_path = "/content/real_mic_test_16k.wav"

# Load same format used during training
y, sr = librosa.load(audio_path, sr=16000, mono=True)

# MFCC
mfcc = librosa.feature.mfcc(
    y=y,
    sr=sr,
    n_mfcc=40
)

# Log-Mel Spectrogram
mel = librosa.feature.melspectrogram(
    y=y,
    sr=sr,
    n_mels=64
)

log_mel = librosa.power_to_db(mel, ref=np.max)

# Same mean + std pooling used for training
features = np.concatenate([
    mfcc.mean(axis=1),
    mfcc.std(axis=1),
    log_mel.mean(axis=1),
    log_mel.std(axis=1)
])

print("===== REAL AUDIO FEATURES =====")
print("Sample rate :", sr)
print("Duration    :", round(len(y) / sr, 2), "seconds")
print("Feature shape:", features.shape)

assert features.shape[0] == 208, "❌ Feature count is not 208!"

print("✅ 208 features extracted successfully!")
```

```python
import torch
import torch.nn as nn

# Same architecture used during audio model training
class AudioBehaviourNet(nn.Module):
    def __init__(self, input_size=208):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(input_size, 128),
            nn.ReLU(),
            nn.Dropout(0.30),

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.25),

            nn.Linear(64, 2)
        )

    def forward(self, x):
        return self.network(x)

print("✅ AudioBehaviourNet defined successfully!")
```
