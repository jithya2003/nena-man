```python
# STEP 4 — Check why PInSoRo frustration evidence has 0 rows

import pandas as pd
import os

base_path = "/content/drive/MyDrive/Behavior Detection/Behavioral_State_Model"

features_file = os.path.join(
    base_path,
    "pinsoRo_features.csv"
)

frustration_file = os.path.join(
    base_path,
    "pinsoRo_frustrated_interval_features.csv"
)

features_df = pd.read_csv(features_file)
frustration_df = pd.read_csv(frustration_file)

print("PInSoRo feature sessions:", len(features_df))
print("PInSoRo frustration rows:", len(frustration_df))

print("\nFirst feature session IDs:")
print(features_df["session_id"].head(10).tolist())

print("\nFirst frustration session IDs:")
if len(frustration_df) > 0:
    print(frustration_df["session_id"].head(10).tolist())
else:
    print("NO ROWS")

# Check whether session IDs overlap
if len(frustration_df) > 0:
    feature_ids = set(features_df["session_id"].astype(str))
    frustration_ids = set(frustration_df["session_id"].astype(str))

    overlap = feature_ids.intersection(frustration_ids)

    print("\nMatching session IDs:", len(overlap))

    if len(overlap) > 0:
        print("Example matching IDs:")
        print(list(overlap)[:10])
else:
    print("\nCannot check overlap because frustration file is empty.")
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

N = 2000

states = [
    "Engaged",
    "Distracted",
    "Frustrated",
    "Overwhelmed",
    "Overloaded"
]

rows = []

for state in states:

    n = N // 5

    if state == "Engaged":
        reading_accuracy = np.random.uniform(0.85, 1.00, n)
        reading_speed = np.random.uniform(1.5, 3.0, n)
        hesitation_count = np.random.randint(0, 2, n)
        pause_duration = np.random.uniform(0.2, 1.0, n)
        fluency_score = np.random.uniform(0.80, 1.00, n)
        error_count = np.random.randint(0, 2, n)
        repeated_errors = np.random.randint(0, 1, n)
        response_latency = np.random.uniform(0.5, 2.0, n)
        retry_count = np.random.randint(0, 2, n)
        skip_count = np.random.randint(0, 1, n)
        idle_time = np.random.uniform(0, 2, n)
        task_completion_rate = np.random.uniform(0.85, 1.00, n)
        gaze_away_frequency = np.random.uniform(0, 2, n)
        attention_duration = np.random.uniform(0.80, 1.00, n)
        head_pose_variation = np.random.uniform(1, 8, n)
        face_visibility = np.random.uniform(0.90, 1.00, n)

    elif state == "Distracted":
        reading_accuracy = np.random.uniform(0.60, 0.90, n)
        reading_speed = np.random.uniform(1.0, 2.5, n)
        hesitation_count = np.random.randint(1, 5, n)
        pause_duration = np.random.uniform(1.0, 3.0, n)
        fluency_score = np.random.uniform(0.50, 0.85, n)
        error_count = np.random.randint(1, 5, n)
        repeated_errors = np.random.randint(0, 3, n)
        response_latency = np.random.uniform(1.5, 4.0, n)
        retry_count = np.random.randint(0, 3, n)
        skip_count = np.random.randint(1, 5, n)
        idle_time = np.random.uniform(3, 8, n)
        task_completion_rate = np.random.uniform(0.50, 0.85, n)
        gaze_away_frequency = np.random.uniform(3, 10, n)
        attention_duration = np.random.uniform(0.30, 0.65, n)
        head_pose_variation = np.random.uniform(5, 20, n)
        face_visibility = np.random.uniform(0.70, 1.00, n)

    elif state == "Frustrated":
        reading_accuracy = np.random.uniform(0.40, 0.75, n)
        reading_speed = np.random.uniform(0.6, 2.0, n)
        hesitation_count = np.random.randint(3, 9, n)
        pause_duration = np.random.uniform(2.0, 6.0, n)
        fluency_score = np.random.uniform(0.30, 0.70, n)
        error_count = np.random.randint(3, 9, n)
        repeated_errors = np.random.randint(2, 6, n)
        response_latency = np.random.uniform(3.0, 7.0, n)
        retry_count = np.random.randint(2, 7, n)
        skip_count = np.random.randint(0, 4, n)
        idle_time = np.random.uniform(2, 7, n)
        task_completion_rate = np.random.uniform(0.30, 0.75, n)
        gaze_away_frequency = np.random.uniform(2, 8, n)
        attention_duration = np.random.uniform(0.30, 0.70, n)
        head_pose_variation = np.random.uniform(8, 25, n)
        face_visibility = np.random.uniform(0.70, 1.00, n)

    elif state == "Overwhelmed":
        reading_accuracy = np.random.uniform(0.25, 0.60, n)
        reading_speed = np.random.uniform(0.3, 1.3, n)
        hesitation_count = np.random.randint(5, 12, n)
        pause_duration = np.random.uniform(4.0, 10.0, n)
        fluency_score = np.random.uniform(0.20, 0.55, n)
        error_count = np.random.randint(4, 10, n)
        repeated_errors = np.random.randint(3, 8, n)
        response_latency = np.random.uniform(5.0, 12.0, n)
        retry_count = np.random.randint(3, 8, n)
        skip_count = np.random.randint(1, 5, n)
        idle_time = np.random.uniform(5, 12, n)
        task_completion_rate = np.random.uniform(0.15, 0.55, n)
        gaze_away_frequency = np.random.uniform(4, 10, n)
        attention_duration = np.random.uniform(0.15, 0.50, n)
        head_pose_variation = np.random.uniform(10, 30, n)
        face_visibility = np.random.uniform(0.60, 1.00, n)

    else:  # Overloaded
        reading_accuracy = np.random.uniform(0.20, 0.55, n)
        reading_speed = np.random.uniform(0.2, 1.1, n)
        hesitation_count = np.random.randint(6, 14, n)
        pause_duration = np.random.uniform(5.0, 12.0, n)
        fluency_score = np.random.uniform(0.15, 0.50, n)
        error_count = np.random.randint(5, 12, n)
        repeated_errors = np.random.randint(4, 10, n)
        response_latency = np.random.uniform(6.0, 15.0, n)
        retry_count = np.random.randint(4, 10, n)
        skip_count = np.random.randint(2, 7, n)
        idle_time = np.random.uniform(7, 15, n)
        task_completion_rate = np.random.uniform(0.10, 0.45, n)
        gaze_away_frequency = np.random.uniform(5, 12, n)
        attention_duration = np.random.uniform(0.10, 0.40, n)
        head_pose_variation = np.random.uniform(12, 35, n)
        face_visibility = np.random.uniform(0.50, 1.00, n)

    for i in range(n):
        rows.append([
            reading_accuracy[i],
            reading_speed[i],
            hesitation_count[i],
            pause_duration[i],
            fluency_score[i],
            error_count[i],
            repeated_errors[i],
            response_latency[i],
            retry_count[i],
            skip_count[i],
            idle_time[i],
            task_completion_rate[i],
            gaze_away_frequency[i],
            attention_duration[i],
            head_pose_variation[i],
            face_visibility[i],
            state
        ])

columns = [
    "reading_accuracy",
    "reading_speed",
    "hesitation_count",
    "pause_duration",
    "fluency_score",
    "error_count",
    "repeated_errors",
    "response_latency",
    "retry_count",
    "skip_count",
    "idle_time",
    "task_completion_rate",
    "gaze_away_frequency",
    "attention_duration",
    "head_pose_variation",
    "face_visibility",
    "behavior_state"
]

behavioral_train_df = pd.DataFrame(rows, columns=columns)

print("Dataset shape:", behavioral_train_df.shape)
print()
print("Class distribution:")
print(behavioral_train_df["behavior_state"].value_counts())
display(behavioral_train_df.head())
```
